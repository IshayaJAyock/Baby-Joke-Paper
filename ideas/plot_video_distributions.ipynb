```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from moviepy.editor import VideoFileClip
from typing import List, Tuple

def plot_video_length_distribution(video_folder: str, data_file: str, output_file: str = 'video_length_distribution.png') -> None:
    """
    Plot the distribution of video lengths from a folder of videos.

    Args:
    video_folder (str): Path to the folder containing video files.
    data_file (str): Path to the Excel file containing video IDs.
    output_file (str): Path to save the output plot (default: 'video_length_distribution.png').

    Returns:
    None
    """
    try:
        # Read the Excel file
        df = pd.read_excel(data_file)
        
        # Get video lengths
        video_lengths = []
        for video_id in df['VideoID']:
            video_path = os.path.join(video_folder, video_id)
            if os.path.exists(video_path):
                with VideoFileClip(video_path) as clip:
                    video_lengths.append(clip.duration)
            else:
                print(f"Warning: Video file not found: {video_path}")
        
        # Plot distribution
        plt.figure(figsize=(10, 6))
        plt.hist(video_lengths, bins=20, edgecolor='black')
        plt.title('Distribution of Video Lengths')
        plt.xlabel('Duration (seconds)')
        plt.ylabel('Frequency')
        plt.savefig(output_file)
        plt.close()
        
        print(f"Plot saved as {output_file}")
    
    except Exception as e:
        print(f"An error occurred: {str(e)}")

def filter_videos_by_length(video_folder: str, data_file: str, min_length: float) -> Tuple[List[str], List[str]]:
    """
    Filter videos based on a minimum length threshold.

    Args:
    video_folder (str): Path to the folder containing video files.
    data_file (str): Path to the Excel file containing video IDs.
    min_length (float): Minimum length threshold in seconds.

    Returns:
    Tuple[List[str], List[str]]: A tuple containing two lists:
        1. List of video IDs that meet the length criteria.
        2. List of video IDs that do not meet the length criteria.
    """
    try:
        # Read the Excel file
        df = pd.read_excel(data_file)
        
        valid_videos = []
        invalid_videos = []
        
        for video_id in df['VideoID']:
            video_path = os.path.join(video_folder, video_id)
            if os.path.exists(video_path):
                with VideoFileClip(video_path) as clip:
                    if clip.duration >= min_length:
                        valid_videos.append(video_id)
                    else:
                        invalid_videos.append(video_id)
            else:
                print(f"Warning: Video file not found: {video_path}")
                invalid_videos.append(video_id)
        
        return valid_videos, invalid_videos
    
    except Exception as e:
        print(f"An error occurred: {str(e)}")
        return [], []

# Example usage:
if __name__ == "__main__":

    VIDEO_FOLDER = 'data/videos'
    DATA_FILE = 'data/excel.xlsx'
    
    
    # Plot video length distribution
    plot_video_length_distribution(VIDEO_FOLDER, DATA_FILE)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from moviepy.editor import VideoFileClip
from typing import List, Tuple
import shutil

def plot_video_length_distribution(video_folder: str, data_file: str, output_file: str = 'video_length_distribution2.png') -> None:
    """
    Plot the distribution of video lengths from a folder of videos.

    Args:
    video_folder (str): Path to the folder containing video files.
    data_file (str): Path to the Excel file containing video IDs.
    output_file (str): Path to save the output plot (default: 'video_length_distribution.png').

    Returns:
    None
    """
    try:
        # Read the Excel file
        df = pd.read_excel(data_file)
        
        # Get video lengths
        video_lengths = []
        for video_id in df['VideoID']:
            video_path = os.path.join(video_folder, video_id)
            if os.path.exists(video_path):
                with VideoFileClip(video_path) as clip:
                    duration = clip.duration
                    video_lengths.append(duration)
                    
                    # Create a subfolder based on the exact length of the video
                    length_folder = os.path.join('video_length', f"{duration:.2f}s")
                    os.makedirs(length_folder, exist_ok=True)
                    
                    # Copy video to the respective folder
                    shutil.copy(video_path, length_folder)
            else:
                print(f"Warning: Video file not found: {video_path}")
        
        # Plot distribution as a bar plot
        length_counts = pd.Series(video_lengths).value_counts().sort_index()
        
        plt.figure(figsize=(10, 6))
        length_counts.plot(kind='bar', color='skyblue', edgecolor='black')
        plt.title('Distribution of Video Lengths')
        plt.xlabel('Duration (seconds)')
        plt.ylabel('Frequency')
        plt.xticks(rotation=45)
        plt.savefig(output_file)
        plt.close()
        
        print(f"Plot saved as {output_file}")
    
    except Exception as e:
        print(f"An error occurred: {str(e)}")

def filter_videos_by_length(video_folder: str, data_file: str, min_length: float) -> Tuple[List[str], List[str]]:
    """
    Filter videos based on a minimum length threshold.

    Args:
    video_folder (str): Path to the folder containing video files.
    data_file (str): Path to the Excel file containing video IDs.
    min_length (float): Minimum length threshold in seconds.

    Returns:
    Tuple[List[str], List[str]]: A tuple containing two lists:
        1. List of video IDs that meet the length criteria.
        2. List of video IDs that do not meet the length criteria.
    """
    try:
        # Read the Excel file
        df = pd.read_excel(data_file)
        
        valid_videos = []
        invalid_videos = []
        
        for video_id in df['VideoID']:
            video_path = os.path.join(video_folder, video_id)
            if os.path.exists(video_path):
                with VideoFileClip(video_path) as clip:
                    if clip.duration >= min_length:
                        valid_videos.append(video_id)
                    else:
                        invalid_videos.append(video_id)
            else:
                print(f"Warning: Video file not found: {video_path}")
                invalid_videos.append(video_id)
        
        return valid_videos, invalid_videos
    
    except Exception as e:
        print(f"An error occurred: {str(e)}")
        return [], []

# Example usage:
if __name__ == "__main__":

    VIDEO_FOLDER = 'data/videos'
    DATA_FILE = 'data/excel.xlsx'
    
    # Plot video length distribution
    plot_video_length_distribution(VIDEO_FOLDER, DATA_FILE)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from moviepy.editor import VideoFileClip
from typing import List, Tuple
import shutil

def plot_video_length_distribution(video_folder: str, data_file: str, x1: float, x2: float, output_file: str = 'video_length_distribution.png') -> None:
    """
    Plot the distribution of video lengths from a folder of videos and copy videos within a specified length range.

    Args:
    video_folder (str): Path to the folder containing video files.
    data_file (str): Path to the Excel file containing video IDs.
    x1 (float): Minimum length threshold in seconds.
    x2 (float): Maximum length threshold in seconds.
    output_file (str): Path to save the output plot (default: 'video_length_distribution.png').

    Returns:
    None
    """
    try:
        # Read the Excel file
        df = pd.read_excel(data_file)
        
        # Create a subfolder for the specified length range
        length_range_folder = f"video_length/{x1}_{x2}"
        os.makedirs(length_range_folder, exist_ok=True)

        # Get video lengths and copy videos within the specified range
        video_lengths = []
        for video_id in df['VideoID']:
            video_path = os.path.join(video_folder, video_id)
            if os.path.exists(video_path):
                with VideoFileClip(video_path) as clip:
                    duration = clip.duration
                    video_lengths.append(duration)
                    
                    # Copy video to the respective folder if within the specified range
                    if x1 <= duration <= x2:
                        shutil.copy(video_path, length_range_folder)
            else:
                print(f"Warning: Video file not found: {video_path}")
        
        # Plot distribution as a bar plot
        length_counts = pd.Series(video_lengths).value_counts().sort_index()
        
        plt.figure(figsize=(10, 6))
        length_counts.plot(kind='bar', color='skyblue', edgecolor='black')
        plt.title('Distribution of Video Lengths')
        plt.xlabel('Duration (seconds)')
        plt.ylabel('Frequency')
        plt.xticks(rotation=45)
        plt.savefig(output_file)
        plt.close()
        
        print(f"Plot saved as {output_file}")
    
    except Exception as e:
        print(f"An error occurred: {str(e)}")

def filter_videos_by_length(video_folder: str, data_file: str, min_length: float) -> Tuple[List[str], List[str]]:
    """
    Filter videos based on a minimum length threshold.

    Args:
    video_folder (str): Path to the folder containing video files.
    data_file (str): Path to the Excel file containing video IDs.
    min_length (float): Minimum length threshold in seconds.

    Returns:
    Tuple[List[str], List[str]]: A tuple containing two lists:
        1. List of video IDs that meet the length criteria.
        2. List of video IDs that do not meet the length criteria.
    """
    try:
        # Read the Excel file
        df = pd.read_excel(data_file)
        
        valid_videos = []
        invalid_videos = []
        
        for video_id in df['VideoID']:
            video_path = os.path.join(video_folder, video_id)
            if os.path.exists(video_path):
                with VideoFileClip(video_path) as clip:
                    if clip.duration >= min_length:
                        valid_videos.append(video_id)
                    else:
                        invalid_videos.append(video_id)
            else:
                print(f"Warning: Video file not found: {video_path}")
                invalid_videos.append(video_id)
        
        return valid_videos, invalid_videos
    
    except Exception as e:
        print(f"An error occurred: {str(e)}")
        return [], []

# Example usage:
if __name__ == "__main__":

    VIDEO_FOLDER = 'data/videos'
    DATA_FILE = 'data/excel.xlsx'
    
    # Define length range
    x1 = 15  # Minimum length in seconds
    x2 = 25  # Maximum length in seconds
    
    # Plot video length distribution and copy videos within the specified range
    plot_video_length_distribution(VIDEO_FOLDER, DATA_FILE, x1, x2)
```
